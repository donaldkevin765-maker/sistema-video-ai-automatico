**Prompt**: inserisci qui sotto quello che vuoi generare.

```python
# @title Prompt Video
PROMPT = "una borsa in pelle su sfondo bianco, luce naturale, qualita professionale"  # @param {type:"string"}
WIDTH = 1000  # @param {type:"integer"}
HEIGHT = 1000  # @param {type:"integer"}
DURATION = 5  # @param {type:"integer"}
```

```python
import torch
from diffusers import CogVideoXPipeline
from diffusers.utils import export_to_video

device = 'cuda' if torch.cuda.is_available() else 'cpu'
print(f'Device: {device}')

pipe = CogVideoXPipeline.from_pretrained(
    'THUDM/CogVideoX-5B',
    torch_dtype=torch.float16 if device == 'cuda' else torch.float32,
)
pipe.to(device)
pipe.enable_model_cpu_offload()
print('Modello caricato!')
```

```python
print(f'Generazione video in corso...\nPrompt: {PROMPT}')

video_frames = pipe(
    prompt=PROMPT,
    num_frames=DURATION * 8,
    width=WIDTH,
    height=HEIGHT,
    guidance_scale=7,
    num_inference_steps=50,
    generator=torch.Generator(device=device).manual_seed(42),
).frames[0]

OUTPUT = f'/content/video-{int(__import__("time").time())}.mp4'
export_to_video(video_frames, OUTPUT, fps=8)
print(f'Video salvato: {OUTPUT}')
```

```python
# Scarica il video
from google.colab import files
files.download(OUTPUT)
print('Download completato!')
```
